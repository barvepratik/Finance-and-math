```python
# Generating and plotting global and rolling standardized time series with trend and seasonality
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Set style for better aesthetics
plt.style.use('seaborn-v0_8')

# Generate synthetic time series data
np.random.seed(42)
n = 1000
t = np.arange(n)
trend = 0.01 * t  # Linear increasing trend
seasonality = 0.5 * np.sin(2 * np.pi * t / 50)  # Sinusoidal component with period 50
noise = np.random.normal(0, 0.2, n)  # Random noise
data = trend + seasonality + noise

# Global standardization
global_mean = np.mean(data)
global_std = np.std(data)
z_global = (data - global_mean) / global_std

# Rolling standardization with window size 20
rolling_mean = pd.Series(data).rolling(window=20, center=True).mean()
rolling_std = pd.Series(data).rolling(window=20, center=True).std()
z_rolling = (pd.Series(data) - rolling_mean) / rolling_std

# Plotting
fig, axs = plt.subplots(3, 1, figsize=(12, 8), sharex=True)

# Plot actual data
axs[0].plot(t, data, label='Actual data', color='Orange')
axs[0].set_title('Actual data')
# axs[0].set_ylabel('Actual data')
axs[0].legend()
axs[0].grid(True)


# Plot global standardized data
axs[1].plot(t, z_global, label='Global Z-Score', color='blue')
axs[1].set_title('Global Standardized Time Series')
axs[1].set_ylabel('Z-Score')
axs[1].legend()
axs[1].grid(True)

# Plot rolling standardized data
axs[2].plot(t, z_rolling, label='Rolling Z-Score (window=20)', color='green')
axs[2].set_title('Rolling Standardized Time Series (Window=20)')
axs[2].set_xlabel('Time')
axs[2].set_ylabel('Z-Score')
axs[2].legend()
axs[2].grid(True)




plt.tight_layout()
# output_path = "/mnt/data/standardized_time_series.png"
# plt.savefig(output_path)

print("Generated synthetic time series with trend and seasonality, computed global and rolling z-scores, and saved plot as 'standardized_time_series.png'")
```
